# Indian E-Commerce Sales Analysis

## Overview
End-to-end data analysis project on Indian e-commerce sales data 
containing 200,000+ transactions across 3 connected tables — 
customers, products and sales.

## Tools Used
- Python (Pandas) — Data Cleaning
- MySQL — Data Analysis (12 queries)
- Power BI — Interactive Dashboard

## Dataset
Kaggle - Indian E-Commerce Sales Dataset

## Tables
- **Customers** — 40,000 rows (customer demographics)
- **Products** — 2,000 rows (product catalogue)  
- **Sales** — 200,000 rows (transaction records)

## Project Workflow
1. Data Cleaning → Python
2. Data Analysis → MySQL (JOINs, CTEs, Window Functions)
3. Visualization → Power BI

## clean column names

```python
df.columns= df.columns.str.strip().str.lower().str.replace(' ', '_')
print(df.columns)
```

## clean text

```python
df['gender']= df['gender'].str.strip().str.lower()
df['customer_tier']= df['customer_tier'].str.strip().str.lower()
```

```python
df['date_of_birth'] = pd.to_datetime(df['date_of_birth'])
```
