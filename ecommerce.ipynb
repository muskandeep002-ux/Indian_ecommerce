# Indian E-Commerce Sales Analysis

## Overview
End-to-end data analysis project on Indian e-commerce sales data 
containing 200,000+ transactions across 3 connected tables — 
customers, products and sales.

## Tools Used
- Python (Pandas) — Data Cleaning
- MySQL — Data Analysis (12 queries)
- Power BI — Interactive Dashboard

## Dataset
Kaggle - Indian E-Commerce Sales Dataset

## Tables
- **Customers** — 40,000 rows (customer demographics)
- **Products** — 2,000 rows (product catalogue)  
- **Sales** — 200,000 rows (transaction records)

## Project Workflow
1. Data Cleaning → Python
2. Data Analysis → MySQL (JOINs, CTEs, Window Functions)
3. Visualization → Power BI

## Cleaning sales table

## Loading .csv file

```python
df= pd.read_csv('sales.csv')
```

Sales Table


## check the number of rows and columns 

```python
print(df.shape)
```


## check the datatype

```python
print(df.info())
```


## check the first 10 rows

```python
print(df.head(10))
```

## check the summary statistics 

```python
print(df.describe())
```

## check the last 5 rows

```python
print(df.tail(5))
```

## clean column names

```python
df.columns= df.columns.str.strip().str.lower().str.replace(' ', '_')
print(df.columns)
```


## check for null values

```python
print(df.isnull().sum())
```

## check duplicates

```python
print(df.duplicated().sum())
```

## Correct data types

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['delivery_date'] = pd.to_datetime(df['delivery_date'])
df['order_time'] = pd.to_datetime(df['order_time'], format='%H:%M:%S').dt.time
```

## Null handling

```python
df['shipping_cost'] = df['shipping_cost'].fillna(df['shipping_cost'].median())
df['review_text'] = df['review_text'].fillna('No Review')
df['coupon_code'] = df['coupon_code'].fillna('No Coupon')
df['rating']= df['rating'].fillna(0)
```

## download data

```python
df.to_csv('sales_cleaned.csv', index=False)
print("saved successfully!")
```
